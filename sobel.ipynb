```python
horizont = np.array ([[-1,0,1],
                     [-2,0,2],
                     [-1,0,1]]) /8
vertical = np.array ([[-1,-2,-1],
                     [0,0,0],
                     [1,2,1]]) /8


def sobel_h(img):
    result = np.zeros(img.shape)
    for i in range(1, img.shape[0]-1):
        for j in range (1, img.shape[1]-1):
            region = img[i-1:i+2, j-1:j+2]
            result[i,j] = np.sum (region*horizont)
    return result

def sobel_v(img):
    result = np.zeros(img.shape)
    for i in range(1, img.shape[0]-1):
        for j in range (1, img.shape[1]-1):
            region = img[i-1:i+2, j-1:j+2]
            result[i,j] = np.sum (region*vertical)
    return result
```
